### Same procedure as notebook.ipynb but with validation data to fine tune thresholds

```python
import pandas as pd
df = pd.read_csv("data/german.data-numeric", sep=r"\s+", header = None)
```

```python
from sklearn.model_selection import train_test_split
x = df.drop(columns=24)
y = df[24]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state=42, test_size=0.2, stratify=y)
```

Creating a split from train data, into validation and train data

```python
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, stratify=y_train, random_state=42, test_size = 0.25)
```

### Scale the data
`fit_transform()` only for training data rest only `transform`

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_val_scaled = scaler.transform(x_val)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_val_scaled)
y_pred
```

```python
y_pred_proba = model.predict_proba(x_val_scaled)
y_pred_proba
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_val, y_pred)

cm
```

```python
FP = cm[0, 1]
FN = cm[1, 0]
cost = FP * 1 + FN * 5

print("FP:", FP)
print("FN:", FN)
print("Cost:", cost)
```

# Check thresholds
On validation data this time instead of directly on test data

```python
import numpy as np 

for threshold in np.arange(0.1, 1.0, 0.1):
    y_pred_threshold = np.where(
        y_pred_proba[:,1] >= threshold, 2, 1
    )

    cm = confusion_matrix(y_val, y_pred_threshold)

    fp = cm[0, 1]
    fn = cm[1, 0]

    cost = fp * 1 + fn * 5

    print(f"Threshold: {threshold:.1f} | FP: {fp} | FN: {fn} | Cost: {cost}")
```

```python
test_probabilities = model.predict_proba(x_test_scaled)[:,1]
test_probabilities
```

### Moving on with the picked threshold (0.1)

```python
final_test_preds = np.where(test_probabilities >= 0.1, 2, 1)
cm_test = confusion_matrix(y_test, final_test_preds)
cm_test
```

```python
fp_test = cm_test[0, 1]
fn_test = cm_test[1, 0]
final_cost = fp_test * 1 + fn_test * 5

print(f"Final Exam Cost: {final_cost}")
print(f"False Positives: {fp_test}")
print(f"False Negatives: {fn_test}")
```
