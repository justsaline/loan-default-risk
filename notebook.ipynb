# Import from skikitlearn and train_test_split the data

```python
from sklearn.model_selection import train_test_split
```

```python
# X has only the attributes that define good and bad
# Y has the values that decide whether applicant is good or bad

X = df.drop(columns=24)
Y = df[24]
```

### Train and Inspect
Split the data into 80-20, 80% is used to train your model whereas 20% to test your model (test_size=0.2), also maintain the ratio of good to bad within the test and train data (stratify=Y), can ensure reproduceability using a random seed (random_state=42)

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
```

We can see that the training data is 80% of the original data and that test data is 70% of the original data

```python
print(Y_train.value_counts())
print(Y_test.value_counts())
```

We can see that the train and test data have split in proper proportions of 7:3 w.r.t the original proportions of the data

## Scale the data
Use StandardScaler to scale data of varying ranges into a understandable and simple range (0-1) and apply it with `.fit` and `.fit_transform` the former for test attributes and the latter for trian attributes.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Check out the scaled data for reference

print(X_train_scaled.shape)
print(X_test_scaled.shape)
print(X_train_scaled[:2])
```

# Choose a model and fit it to your data
Logistic Regression here is used because we are trying to predict the class of applicants (Good or Bad), it predicts the probablity of good or bad and assigns a class to each row

```python
from sklearn.linear_model import LogisticRegression

# set the model to LogisticRegression
model = LogisticRegression()

# Train the model on train data X_train and its results Y_train
model.fit(X_train_scaled, Y_train)
```

Now that the model is trained test on your test_data to check how it performs
use the `.predict` function on scaled X_test data to check the results

```python
y_pred = model.predict(X_test_scaled)
```

```python
print(y_pred[:20])
print(Y_test.iloc[:20].values)
```

We can see here that the predicted data is a little varying to the actual outcomes that shouldve come

# Check out the metrics
Check out the accuracy of the model the confusion matrix, precision, recall and F1

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test, y_pred)

print(accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test, y_pred)

print(cm)

# Use this to get the TP, TN, FP, FN
# here it is 29, 125, 15, 31 respectively
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(Y_test, y_pred, pos_label=2)
recall = recall_score(Y_test, y_pred, pos_label=2)

print("Precision:", precision)
print("Recall:", recall)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(Y_test, y_pred, pos_label=2)

print("F1:", f1)
```

### Check the probability of good and bad for each applicant in the data

0th Column - Probability of Good

1st Column - Probability of Bad

```python
y_proba = model.predict_proba(X_test_scaled)
print(y_proba)
print(model.classes_)
```

```python
bad_probability = model.predict_proba(X_test_scaled)[:, 1]
bad_probability
```

# Check the precision and recall with different Thresholds

```python
import numpy as np

threshold = 0.4
y_pred_04 = np.where(bad_probability >= threshold, 2,1)
```

```python
precision_04 = precision_score(Y_test, y_pred_04, pos_label=2)
recall_04 = recall_score(Y_test, y_pred_04, pos_label=2)

print("Precision:", precision_04)
print("Recall:", recall_04)
```

### Loop through each threshold from 0.1 to 1.0 and check the precision and recall

```python
thresholds = np.arange(0.1, 1.0, 0.1)

for threshold in thresholds:
    y_pred_threshold = np.where(bad_probability >= threshold, 2, 1)

    precision = precision_score(Y_test, y_pred_threshold, pos_label=2)
    recall = recall_score(Y_test, y_pred_threshold, pos_label=2)

    print(
        f"Threshold: {threshold:.1f} | "
        f"Precision: {precision:.2f} | "
        f"Recall: {recall:.2f}"
    )
```

## Find the actual cost of FP and FN as both are not equal
A FN is worth 5 times a FP

```python
cm = confusion_matrix(Y_test, y_pred_04)
FP = cm[0, 1]
FN = cm[1, 0]
print(cm)
```

```python
cost = FP * 1 + FN * 5

print("FP:", FP)
print("FN:", FN)
print("Cost:", cost)
```

### Cost for each threshold in the test data

```python
for threshold in thresholds:
    y_pred_threshold = np.where(
        bad_probability >= threshold, 2, 1
    )

    cm = confusion_matrix(Y_test, y_pred_threshold)

    fp = cm[0, 1]
    fn = cm[1, 0]

    cost = fp * 1 + fn * 5

    print(f"Threshold: {threshold:.1f} | FP: {fp} | FN: {fn} | Cost: {cost}")
```

### Conclusion
Since we tested the thresholds directly on the test data we cannot be sure that it might hold true for all of the credit data, therefore creating a validation data (splitting test data) would be a better approach while analysing threasholds, which is what we'll do in our other notebook, and even though threshold 0.3 is the least expensive we can't be for sure it is the right one.
